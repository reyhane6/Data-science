```python
import math
class Circle():
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return round(math.pi*self.radius**2,2)
  def set_radius(self,new_radius):
    self.radius=new_radius
```

```python
c=Circle(5)
print(c.area())
c.set_radius(10)
print(c.area())
```

```python
class Student():
  def __init__(self,ad,qiymet):
    self.ad=ad
    self.qiymet=qiymet
  def average(self):
    return sum(self.qiymet)/len(self.qiymet)
```

```python
class BankAccount():
  def __init__(self,balans):
    self.balans=balans
  def deposit(self,amount):
    self.balans+=amount
  def withdraw(self,amount):
    self.balans-=amount
  def get_balance(self):
    return self.balans
```

```python
class Library():
  def __init__(self):
    self.books=[]
  def add_book(self,book):
    self.books.append(book)
  def list_books(self):
    for book in self.books:
      book.info()
  def find_by_author(self,author):
    found_books=[]
    for book in self.books:
      if book.author.lower()==author.lower():
        found_books.append(book)
    return found_books
  def find_by_year(self,year):
    found_books=[]
    for book in self.books:
      if book.year==year:
        found_books.append(book)
    return found_books
  def remove_book(self,title):
    for book in self.books:
      if book.title.lower()==title.lower():
        self.books.remove(book)
        return f'{title} silindi'
    print('kitab tapilmadi')
  def search_by_author(self,author_name):
    found_books=[]
    for book in self.books:
      if book.author.lower()==author_name.lower():
        found_books.append(book)
    return found_books
```

```python
def longest_increasing_subarrey(lst):
  max_seq=[]
  current=[lst[0]]
  for i in range(1,len(lst)):
    if lst[i]>lst[i-1]:
      current.append(lst[i])
    else:
      if len(current)>len(max_seq):
        max_seq=current
      current=[lst[i]]
  if len(current)>len(max_seq):
    max_seq=current
  return max_seq
print(longest_increasing_subarrey([1,2,3,1,2,3,4]))
```
