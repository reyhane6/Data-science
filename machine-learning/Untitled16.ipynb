```python
for col in df.columns:
  plt.figure(figsize=(10,6))
  sns.histplot(df[col],kde=True)
  plt.show()
```

```python
for col in df.columns:
  plt.figure(figsize=(10,6))
  sns.boxplot(x=df[col])
  plt.show()
```

```python
for col in df.columns:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outlier=df[(df[col]<lower)|(df[col]>upper)]
  outlier
```
